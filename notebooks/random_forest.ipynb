## Introduction

This notebook explores the use of **Random Forest** as a classification model for pneumonia detection from chest X-ray images.

### Why Random Forest?

After reaching a performance plateau with Logistic Regression (best F1: 0.838, Test AUC: 0.904), a more expressive model is needed to capture non-linear patterns in the image data.

Random Forest was chosen because:

- It is an **ensemble method** that combines multiple decision trees, reducing variance and overfitting compared to a single tree
- It handles **high-dimensional data** well (4096 features from 64x64 images) without requiring PCA
- It natively supports **class_weight='balanced'** to handle the dataset imbalance
- It provides **feature importance** scores, which can help understand which pixel regions matter most
- It serves as a natural next step before exploring deep learning approaches
